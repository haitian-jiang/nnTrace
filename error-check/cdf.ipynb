```python
import numpy as np
from math import sqrt
import matplotlib.pyplot as plt
```

```python
def inv_cdf(u):
    if 0 <= u < 1/8:
        return -1-2*u+2*sqrt(u*(u+1))
    elif 1/8 <= u < 7/8:
        return 4/3*u-2/3
    elif 7/8 <= u <= 1:
        return 3-2*u-sqrt((2*u-3)**2-1)
    else:
        raise ValueError('u must be in [0,1]')
```

```python
x = np.random.rand(100000)
y = np.array([inv_cdf(u) for u in x])
```

```python
plt.hist(y, bins=1000, density=True)
plt.show()
```

```python
y2 = y[:, np.newaxis] + y[np.newaxis, :]
```

```python
y2 = y2.flatten()
```

```python
plt.hist(y2, bins=1000, density=True)
plt.show()
```

```python
xx = np.random.rand(100000)
delta3 = np.array([inv_cdf(u) for u in xx])
```

```python
y3 = y2.reshape(100000, 100000) + delta3[:, np.newaxis]
```

```python
y3 = y3.flatten()
plt.hist(y3, bins=1000, density=True)
plt.show()
```
